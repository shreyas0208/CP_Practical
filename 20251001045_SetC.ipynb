I am btech 2nd year chemical engineering student, reaction equilibrium of the reaction A+B->C, initial moles of A=B=0.5 and that of C=0. If x moles of C is formed such that Kc=x/(0.5-x^2) which makes the quadratic equation Kc(x^2)-(Kc-1)x+0.25Kc=0 write a python program that readsKc using input, prints the data type of the raw input and of the converted valueand shows a clear message if the text is not a number or if Kc<=0, finds a,b,c the discriminant and both the roots, selects the physically meaningful root (between 0 and 0.5), prints  concentration of A,B,C upto 3 decimal places and conversion of limiting reactant (x/ 0.5)*100(%), classifies the conversion using if,elif,else >=70% "High",>=40% "Moderate", otherwise low, do not use advanced coding, keep it simple

```python
import math

raw_input_value = input("Enter Kc: ")
print("Data type of raw input:", type(raw_input_value))

try:
    Kc = float(raw_input_value)
except ValueError:
    print("Error: Please enter a number.")
else:
    print("Data type of converted value:", type(Kc))

    if Kc <= 0:
        print("Error: Kc must be greater than 0.")
    else:
        # Coefficients of Kc*x^2 - (Kc+1)*x + 0.25*Kc = 0
        a = Kc
        b = -(Kc + 1)
        c = 0.25 * Kc

        discriminant = b**2 - 4 * a * c
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)

        print("a =", a)
        print("b =", b)
        print("c =", c)
        print("Discriminant =", discriminant)
        print("Root 1 =", root1)
        print("Root 2 =", root2)

        if 0 < root1 < 0.5:
            x = root1
        else:
            x = root2

        concentration_A = 0.5 - x
        concentration_B = 0.5 - x
        concentration_C = x
        conversion = (x / 0.5) * 100

        print(f"Physically meaningful root, x = {x:.3f}")
        print(f"Equilibrium concentration of A = {concentration_A:.3f}mol/l")
        print(f"Equilibrium concentration of B = {concentration_B:.3f}mol/l")
        print(f"Equilibrium concentration of C = {concentration_C:.3f}mol/l")
        print(f"Conversion of limiting reactant = {conversion:.3f}%")

        if conversion >= 70:
            print("Conversion classification: High")
        elif conversion >= 40:
            print("Conversion classification: Moderate")
        else:
            print("Conversion classification: Low")
```

#Question 2 Part A

I am btech 2nd year chemical engineering student, dc/dt=-kC.Use the Euler method C_new = C_old -kC old Δt. Take Co=1 mol/l, k=0.12(min)^-1, Δt=0,5 min. Using a for loop simulate 5min and print a table of time (min), concentration (mol/l) and conversion (%), using math.exp(), calculate the exact value of Coe^(-kt) at 5 mnand print the absolute error of the euler result, using a while loop, repeat the simulation until the conversion reaches 70%or time reaches 100 min. Print the time needed and the final concentration, do not use advanced coding, keeep it simple

```python
import math

Co = 1.0       # Initial concentration (mol/L)
k = 0.12       # Rate constant (1/min)
dt = 0.5       # Time step (min)

# Simulate from 0 to 5 minutes using Euler's method
C = Co
time = 0.0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")
print(f"{time:.1f}\t\t{C:.4f}\t\t\t{(Co - C) / Co * 100:.2f}")

for step in range(1, 11):
    C = C - k * C * dt
    time = step * dt
    conversion = (Co - C) / Co * 100
    print(f"{time:.1f}\t\t{C:.4f}\t\t\t{conversion:.2f}")

# Calculate the exact concentration at 5 minutes
exact_C = Co * math.exp(-k * 5)
absolute_error = abs(C - exact_C)

print(f"\nExact concentration at 5 min: {exact_C:.4f} mol/L")
print(f"Absolute error at 5 min: {absolute_error:.4f} mol/L")

# Continue with Euler's method until conversion reaches 70% or time reaches 100 min
C = Co
time = 0.0
conversion = 0.0

while conversion < 70 and time < 100:
    C = C - k * C * dt
    time = time + dt
    conversion = (Co - C) / Co * 100

print(f"\nTime reached: {time:.1f} min")
print(f"Final concentration: {C:.4f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```

#Question 2 Part B

 
I am btech 2nd year chemical engineering student, write a program in python that reads a lot code, loops through each character and prints the number of digits and the sum of the digits. Show a message for an empty code and for  acode with no digits

```python
lot_code = input("Enter the lot code: ")

if lot_code == "":
    print("The lot code is empty.")
else:
    digit_count = 0
    digit_sum = 0

    for character in lot_code:
        if character in "0123456789":
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The lot code contains no digits.")
    else:
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)
```

#Question 3 Part A

I am btech 2nd year chemical engineering student, write a recursive function factorial (n) remember the base case for 0! and a function combination (n,r) = n!/(r!(n-r)!) that uses it reject the invalid input ( negative values, r>n), print the number of blends for n=8,r=4 and show the message for n=3 r=5, do not use advanced coding keep it simplea

```python
def factorial(n):
    if n == 0:          # Base case: 0! = 1
        return 1
    else:
        return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0 or r > n:
        return None     # Invalid input
    else:
        return factorial(n) // (factorial(r) * factorial(n - r))


n = 8
r = 4
result = combination(n, r)

if result is not None:
    print(f"Number of blends for n={n}, r={r}: {result}")
else:
    print("Invalid input.")

n = 3
r = 5
result = combination(n, r)

if result is not None:
    print(f"Number of blends for n={n}, r={r}: {result}")
else:
    print(f"Invalid input: r ({r}) cannot be greater than n ({n}).")
```

#Question 3 Part B

I am btech 2nd year chemical engineering student, during a cooling experiment temperature readings are taken at times = Fibonacci number * 3min. Write an iterative function fib_iter(n) that returns a list of the first n fibonacci numbers and a recursive function fib_rec(k) that returns the kth fibonacci number(Fo=1,F1=1). For n =8, check that both functions agree and print a table: Reading No. I Fibonacci tag I Time(min), do not use advanced coding, keep it simple

```python
def fib_iter(n):
    numbers = []

    for i in range(n):
        if i == 0 or i == 1:
            numbers.append(1)
        else:
            numbers.append(numbers[i - 1] + numbers[i - 2])

    return numbers


def fib_rec(k):
    if k == 0 or k == 1:
        return 1
    else:
        return fib_rec(k - 1) + fib_rec(k - 2)


n = 8
fibonacci_list = fib_iter(n)

# Check the iterative and recursive functions give the same numbers
agree = True
for k in range(n):
    if fibonacci_list[k] != fib_rec(k):
        agree = False

if agree:
    print("Both functions agree.")
else:
    print("The functions do not agree.")

print("\nReading No.\tFibonacci tag\tTime (min)")
for i in range(n):
    time = fibonacci_list[i] * 3
    print(f"{i + 1}\t\t{fibonacci_list[i]}\t\t{time}")
```

#Question 4

I am btech 2nd year chemical engineering student,photocatalytic degradation of a dye was studied at 2 UV intensities. The conc(mol/L) was recorded: Time_min : 0,10,20,30,40,50; Run1_mol_L(low UV) : 0.8,0.54,0.36,0.24,0.16,0.11 and Run2_mol_L(high UV):  0.8,0.44,0.24,0.13,0.07,0.04. Create a data frame from this table. Save it as kinetics.csv (without the index column) and as kinetics.json. Read the CSV back and print its shape and data types . Print mean and standard deviation of Run 1 and Run 2 and state which run shows more variation. Plot both runs on one graph(Time on x axis) with title, axis labels with units, legends and grid. Air is compressed sloely in apiston cylinder vessel at constant temperature: Volume_L : 2,3,4,5,6 and Pressure_bar 3.01,1.98,1.52,1.19,1.01. Plot pressure vs volume as a scatter plot. Calculate k as the mean of P*V and draw the reference curve P=k/V on the same graph. Add title, labels with units and legend. Do not use advanced coding keep it simple 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# Photocatalytic dye degradation data
kinetics_data = {
    "Time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_L": [0.8, 0.54, 0.36, 0.24, 0.16, 0.11],
    "Run2_mol_L": [0.8, 0.44, 0.24, 0.13, 0.07, 0.04],
}

kinetics_df = pd.DataFrame(kinetics_data)
print("Kinetics data:")
print(kinetics_df)

kinetics_df.to_csv("kinetics.csv", index=False)
kinetics_df.to_json("kinetics.json", orient="records", indent=2)

# Read the CSV file back and inspect its shape and data types.
kinetics_from_csv = pd.read_csv("kinetics.csv")
print("\nCSV shape:", kinetics_from_csv.shape)
print("\nCSV data types:")
print(kinetics_from_csv.dtypes)

# Calculate and compare sample standard deviations.
run1_mean = kinetics_from_csv["Run1_mol_L"].mean()
run1_std = kinetics_from_csv["Run1_mol_L"].std()
run2_mean = kinetics_from_csv["Run2_mol_L"].mean()
run2_std = kinetics_from_csv["Run2_mol_L"].std()

print(f"\nRun 1 mean: {run1_mean:.4f} mol/L")
print(f"Run 1 standard deviation: {run1_std:.4f} mol/L")
print(f"Run 2 mean: {run2_mean:.4f} mol/L")
print(f"Run 2 standard deviation: {run2_std:.4f} mol/L")

if run1_std > run2_std:
    print("Run 1 shows more variation.")
elif run2_std > run1_std:
    print("Run 2 shows more variation.")
else:
    print("Both runs show the same variation.")

# Plot both photocatalytic degradation runs.
plt.figure()
plt.plot(
    kinetics_from_csv["Time_min"],
    kinetics_from_csv["Run1_mol_L"],
    marker="o",
    label="Run 1 (low UV)",
)
plt.plot(
    kinetics_from_csv["Time_min"],
    kinetics_from_csv["Run2_mol_L"],
    marker="s",
    label="Run 2 (high UV)",
)
plt.title("Photocatalytic Dye Degradation")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True)

# Isothermal compression data
pressure_volume_df = pd.DataFrame(
    {
        "Volume_L": [2, 3, 4, 5, 6],
        "Pressure_bar": [3.01, 1.98, 1.52, 1.19, 1.01],
    }
)

pressure_volume_df["PV_bar_L"] = (
    pressure_volume_df["Pressure_bar"] * pressure_volume_df["Volume_L"]
)
k = pressure_volume_df["PV_bar_L"].mean()
print(f"\nMean value of P*V (k): {k:.4f} bar L")

# Draw P = k/V over the measured volume range.
reference_volume = np.linspace(
    pressure_volume_df["Volume_L"].min(),
    pressure_volume_df["Volume_L"].max(),
    100,
)
reference_pressure = k / reference_volume

plt.figure()
plt.scatter(
    pressure_volume_df["Volume_L"],
    pressure_volume_df["Pressure_bar"],
    label="Measured data",
)
plt.plot(reference_volume, reference_pressure, label=f"Reference curve (k={k:.2f} bar L)")
plt.title("Pressure vs. Volume During Isothermal Compression")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)

plt.show()
```
