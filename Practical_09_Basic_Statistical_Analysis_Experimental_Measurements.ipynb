## 2. Chemical Engineering Context

Repeated measurements are common in chemical engineering laboratories. For example, a student may measure:

- temperature several times,
- pressure several times,
- concentration several times,
- pH several times,
- flow rate several times.

The measurements may not be exactly identical because of instrument limitations, environmental conditions and experimental variation.

Two simple statistics help us describe the measurements:

### Mean

The mean gives the **average value** of the measurements.

### Standard deviation

Standard deviation gives an indication of **how much the measurements vary around the mean**.

These statistics do not by themselves prove that an experiment is accurate or scientifically valid; they provide a basic numerical description of the observed measurements.

### Copilot Activity 3 – Standard Deviation

````python
# Activity 4
# Create DataFrames for Set A and Set B and compare their standard deviations.

from typing import Set


std_dv_a = pd.DataFrame({
    "Set A": [78.7, 78.8, 78.7, 78.8, 78.7],
    "Set B": [76.0, 81.5, 78.7, 82.0, 74.8]
})

print(f"Standard Deviation of Set A: {std_dv_a['Set A'].std():.2f}, \
Mean of Set A : {std_dv_a['Set A'].mean():.2f}")

print(f"Standard Deviation of Set B: {std_dv_a['Set B'].std():.2f}, \
Mean of Set B : {std_dv_a['Set B'].mean():.2f}")
````

### Important Pandas point

Pandas `Series.std()` uses **sample standard deviation** by default (`ddof=1`). For this practical, accept the Pandas default unless the instructor specifically asks you to calculate population standard deviation.

Do not confuse the two definitions. The important programming skill here is knowing what statistical quantity your software function is calculating.

````python
# Activity 5
# Calculate standard deviation using Pandas.

print(temperature_data['Temperature_C'].std())
````

## 11. Activity 9 – Grouped Statistical Analysis

Create a dataset with an `Experiment_ID` column and repeated temperature measurements for at least three experiments.

Example structure:

| Experiment_ID | Temperature_C |
|---|---:|
| EXP001 | 78.2 |
| EXP001 | 79.1 |
| EXP001 | 78.7 |
| EXP002 | 81.2 |
| EXP002 | 80.8 |
| ... | ... |

Use Pandas `groupby()` to calculate **mean and standard deviation for each experiment**.

This is a key pattern for real experimental datasets: the same statistical operation can be applied separately to different experimental groups.

````python
# Activity 9
# Create grouped experimental data and calculate mean and standard deviation by Experiment_ID.

df_exper_group = pd.DataFrame({
    'Experiment_ID': ['Exp1', 'Exp1', 'Exp1', 'Exp2', 'Exp2', 'Exp2', 'Exp3', 'Exp3', 'Exp3'],
    'Temperature_C': [78.2, 78.5, 78.7, 79.1, 79.0, 78.9, 78.6, 78.8, 78.7],
    'Measurement_No': [1, 2, 3, 1, 2, 3, 1, 2, 3],
})

df_exper_group
````

````python
df_grouped = df_exper_group.groupby('Experiment_ID').agg({'Temperature_C': ['mean', 'std']})
print(df_grouped)
````

## 15. Final Mini-Project – Experimental Measurement Statistics

Build one clean Pandas-based statistical-analysis workflow.

Your final workflow should:

1. Load experimental measurements from the CSV created in the previous practical, or create a similar experimental dataset.
2. Display the dataset.
3. Identify the numerical measurement columns.
4. Calculate the **mean** for each measurement.
5. Calculate the **standard deviation** for each measurement.
6. Calculate minimum and maximum values.
7. Group the data by an experimental condition or `Experiment_ID`.
8. Calculate mean and standard deviation for each group.
9. Add one unusual measurement and observe its effect.
10. Create one suitable visualization of the variation.
11. Write a short engineering interpretation of the results.

### Required final interpretation

Do not only display numbers. Write 3–5 sentences explaining:

- which measurement has the greatest variation,
- which measurement is relatively consistent,
- what the standard deviation tells you about the repeated measurements,
- whether any unusual observation requires further investigation.

**Do not claim that a small standard deviation automatically means that an experiment is accurate. Precision/variation and accuracy are different concepts.**

````python
# FINAL PROJECT
# Experimental Measurement Statistics

import pandas as pd
import matplotlib.pyplot as plt

# 1. Load and display the data
df = pd.read_csv("chemical_experiment.csv")
display(df)

# 2. Identify the numeric columns and measurements
numeric_columns = df.select_dtypes(include="number").columns
print("Numeric columns:", list(numeric_columns))

measurement_columns = [
    "Temperature_C",
    "Pressure_bar",
    "Concentration_mol_L"
]

# 3. Calculate mean, standard deviation, minimum, and maximum
summary = pd.DataFrame({
    "Mean": df[measurement_columns].mean(),
    "Standard Deviation": df[measurement_columns].std(),
    "Minimum": df[measurement_columns].min(),
    "Maximum": df[measurement_columns].max()
})
display(summary)

# 4. Calculate mean and standard deviation for each experiment
group_summary = df.groupby("Experiment_ID")[measurement_columns].agg(
    ["mean", "std"]
)
display(group_summary)

# 5. Add a simulated unusual concentration reading for EXP002
df_with_unusual = df.copy()
df_with_unusual.loc[len(df_with_unusual)] = [
    "EXP002", 30, 39, 1.14, 2.50
]
display(df_with_unusual.tail())

# Compare concentration statistics before and after adding the reading
print("Original concentration mean:", df["Concentration_mol_L"].mean())
print("New concentration mean:", df_with_unusual["Concentration_mol_L"].mean())
print("Original concentration standard deviation:",
      df["Concentration_mol_L"].std())
print("New concentration standard deviation:",
      df_with_unusual["Concentration_mol_L"].std())

# 6. Plot the concentration variation before and after
plt.boxplot(
    [
        df["Concentration_mol_L"],
        df_with_unusual["Concentration_mol_L"]
    ],
    tick_labels=["Original", "With unusual reading"],
    showmeans=True
)
plt.ylabel("Concentration (mol/L)")
plt.title("Effect of an Unusual Concentration Reading")
plt.show()

# 7. Engineering interpretation
print("Temperature has the largest standard deviation in its own units, ")
print("while pressure is relatively consistent compared with its mean. ")
print("Standard deviation describes how spread out the measurements are; ")
print("because these readings are from different times, the spread can also ")
print("reflect changing process conditions. The unusual concentration reading ")
print("increases the concentration mean and standard deviation. Check it against ")
print("the lab record or instrument readings before deciding how to handle it.")
````
